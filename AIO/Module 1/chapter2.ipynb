# 1. Getting Max Over Kernel

Tìm giá trị lớn nhất trong cửa sổ trượt trên danh sách cho trước

Bài toán Cho một list các số nguyên num_list và một sliding window
có kích thước size k di chuyển từ trái sang phải. Mỗi lần dịch chuyển 1 vị
trí sang phải có thể nhìn thấy đươc k số trong num_list và tìm số lớn nhất
trong k số này sau mỗi lần trượt k phải lớn hơn hoặc bằng 1.

```python
from collections import deque
```

Ý tưởng:
* Tạo deque sao cho phần tử đấu là max
* Nếu deque k còn chỗ thì loại bỏ last element để đưa phần tử khác vào
* Nếu deque còn chỗ thì thêm vào
* Khi thêm vào nếu mà deque 

```python
def get_max_over_kernel(num_list, k=3):
    result = []

    for i in range(len(num_list) - k + 1):

        max_index = i

        for j in range(i, i + k):
            if num_list[max_index] < num_list[j]:
                max_index = j

        result.append(num_list[max_index])

    return result

num_list =  [3, 4, 5, 1,-44 , 5 ,10, 12 ,33, 1]
k = 3
max_list = get_max_over_kernel(num_list, k)
print(max_list)
```

```python
def get_max_over_kernel(num_list, k=3):
    
    dq = deque()
    result = []

    for i in range(len(num_list)):
        if dq and dq[0] <= i - k:
            dq.popleft()
        
        while dq and num_list[dq[-1]] < num_list[i]:
            dq.pop()
        
        dq.append(i)

        if i >= k - 1:
            result.append(num_list[dq[0]])

    return result

num_list =  [3, 4, 5, 1,-44 , 5 ,10, 12 ,33, 1]
k = 3
max_list = get_max_over_kernel(num_list, k)
print(max_list)
```
