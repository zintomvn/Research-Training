# Bài 1

```python
def calc_f1_score(tp, fp, fn):
    if not isinstance(tp, int):
        print("tp must be int")
        return
    if not isinstance(fp, int):
        print("fp must be int")
        return
    if not isinstance(fn, int):
        print("fn must be int")
        return
    elif tp < 0 or fp < 0 or fn < 0:
        print("tp and fp and fn must be greater than or equal zero")
        return
    
    recall = 0
    precision = 0
    f1_score = 0
    if tp + fp != 0:
        precision = tp / (tp + fp)
    else:
        print("(tp + fp) must be greater than 0")
        return 0, 0, 0
    
    if tp + fn != 0:
        recall = tp / (tp + fn)
    else:
        print("(tp + fn) must be greater than 0")
        return 0, 0, 0

    if precision + recall != 0:
        f1_score = 2 * precision * recall / (precision + recall)
    else:
        print("(precision + recall) must be greater than 0")
        return 0, 0, 0
    
    return f1_score

f1_score = calc_f1_score(tp = 2, fp = 4, fn = 5)
print(round(f1_score, 2))
```

# Bài 4

```python
def calc_sig(x):
    return 1 / (1 + math.exp(-x))

assert round(calc_sig(3), 2) == 0.95
print(round(calc_sig(2), 2))
```

# Bài 5

```python
def calc_elu(x, alpha=0.01):
    if not is_number(x):
        print("Input is not number")
        return 0.0
    if x <= 0:
        return alpha * (math.exp(x) - 1)
    if x > 0:
        return x
    return 0.0
```

```python
assert round(calc_elu(1)) == 1
print(round(calc_elu(-1), 2))
```

# Bài 6

```python
def sigmoid(z):
    return 1 / (1 + math.exp(-z))

def relu(z):
    return max(0, z)

def elu(z, alpha=0.01): 
    if z >= 0:
        return alpha * (math.exp(z) - 1)
    return z
```

```python
def calc_activation_func(x, activation_func='sigmoid'):
    if not is_number(x):
        print("x must be a number")
        return 0
    
    x = float(x)
    if activation_func not in ['sigmoid', 'relu', 'elu']:
        print(f"{activation_func} is not supported")
        return 0
    
    if activation_func == 'sigmoid':
        return sigmoid(x)
    if activation_func == 'relu':
        return relu(x)
    if activation_func == 'elu':
        return elu(x, alpha=0.01)
    return 0
```

```python
assert calc_activation_func(x=1, activation_func='relu') == 1
print(round(calc_activation_func(x=3, activation_func='sigmoid'), 2))
```

```python
def calc_ae(y, y_hat):
    if not is_number(y) or not is_number(y):
        print("Input is not number")
        return 0.0 
    return math.fabs(y - y_hat)

y = 1
y_hat = 6
assert calc_ae(y, y_hat) == 5
y = 2
y_hat = 9
print(calc_ae(y, y_hat))
```

# Bài 8

```python
def calc_se(y, y_hat):
    if not is_number(y) or not is_number(y):
        print("Input is not number")
        return 0.0 
    return math.fabs((y - y_hat) ** 2)

y = 4
y_hat = 2
assert calc_se(y, y_hat) == 4
print(calc_se(2, 1))
```

# Bài 9

```python
def approx_sin(x, n):
    if not is_number(x):
        return 0.0
    sum = 0.0
    for i in range(n):
        sum += (-1)**i * x**(2*i + 1) / math.factorial(2*i + 1)
    return sum

assert round(approx_sin(x=1, n=10), 4) == 0.8415
print(round(approx_sin(x=3.14, n=10), 4))
```

# Bài 10

```python
def approx_cos(x, n):
    if not is_number(x):
        return 0.0
    sum = 0.0
    for i in range(n):
        sum += (-1)**i * x**(2*i) / math.factorial(2*i)
    return sum

assert round(approx_cos(x=1, n=10), 2) == 0.54
print(round(approx_cos(x=3.14, n=10), 2))
```

# Bài 11

```python
def approx_sinh(x, n):
    if not is_number(x):
        return 0.0
    sum = 0.0
    for i in range(n):
        sum += x**(2*i + 1) / math.factorial(2*i + 1)
    return sum

assert round(approx_sinh(x=1, n=10), 2) == 1.18
print(round(approx_sinh(x=3.14, n=10), 2))
```

# Bài 12

```python
def approx_cosh(x, n):
    if not is_number(x):
        return 0.0
    sum = 0.0
    for i in range(n):
        sum += x**(2*i) / math.factorial(2*i)
    return sum

assert round(approx_cosh(x=1, n=10), 2) == 1.54
print(round(approx_cosh(x=3.14, n=10), 2))
```
